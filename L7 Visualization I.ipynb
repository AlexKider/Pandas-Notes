### Histograms

- A histogram collects quantitative data **into bins**

- Then plots this binned data

- **Each bin reflects the density** of datapoints with values that lie between the left and right ends of the bin

- The **area (not height!)** of each bin is proportional to the **percentage** of datapoints it contains

```python
# The `edgecolor` argument controls the color of the bin edges
gni = wb["Gross national income per capita, Atlas method: $: 2016"]
plt.hist(gni, density=True, edgecolor="black")

# Add labels
plt.xlabel("Gross national income per capita")
plt.ylabel("Density")
plt.title("Distribution of gross national income per capita");
```

### Interpreting Histograms

- Skewness and Tails

  - Skewed left (negative skew) vs skewed right (positive skew)

  - Most commonly occurring data

**Skewness & Tails**

- It describes in which its "tail" extends
  
  - In a right-skewed distribution, the few large outliers “pull” the mean to the right of the median

- A good way to remember this is that it looks like a “**P**” on its side for **P**ositive skew!

```python
sns.histplot(data = wb, x = 'Gross national income per capita, Atlas method: $: 2016', stat = 'density');
plt.title('Distribution with a long right tail');
```

- An example of long left-tailed below

```python
sns.histplot(data = wb, x = 'Access to an improved water source: % of population: 2015', stat = 'density');
plt.title('Distribution with a long left tail');
```

- In the case where a distribution has equal-sized right and left tails, it is **symmetric**

- **The mean is approximately equal to the median**. Think of mean as the balancing point of the distribution

**Modes (众数, 峰, 模态)**

- The mode is the most frequent value in a distribution

- Can be multiple modes

- The def here is not as strict as stats

- If we set the number of bins to 5, the distribution appears **unimodal** (has one clear peak/mode)

```python
# Rename the very long column name for convenience
wb = wb.rename(columns={'Antiretroviral therapy coverage: % of people living with HIV: 2015':"HIV rate"})
# With 5 bins, it seems that there is only one peak
sns.histplot(data=wb, x="HIV rate", stat="density", bins=5)
plt.title("5 histogram bins");
```

- If a distribution has two peaks/modes, it is called **bimodal**. If a distribution has more peaks than that, we say it is **multimodal**

```python
# With 10 bins, there seem to be two peaks

sns.histplot(data=wb, x="HIV rate", stat="density", bins=10)
plt.title("10 histogram bins");
```

**Outliers**

- In a loose sense, an outlier is a data point that lies an abnormally large distance away from other values
  
- However, in order to make this more concrete, we **define outliers using quartiles**

**Quartiles**

- It **devides data points into four quarters of roughly equal size**

  - **The 1st quartile (Q1)** is the 25th percentile of the distribution; 25% of the data is smaller than or equal to Q1

  - **The 2nd quartile (Q2)** is the 50th percentile of the distribution, or equivalently the median of the distribution. 50% of the data is smaller than or equal to Q2

  - **The 3rd quartile (Q3)** is the 75th percentile of the distribution; 75% of the data is smaller than or equal to Q3

  - The interval **[Q1, Q3]** contains the “middle 50% of the data”

```python
gdp = wb['Gross domestic product: % growth : 2016']
gdp = gdp[~gdp.isna()]

q1, q2, q3 = np.percentile(gdp, [25, 50, 75])

wb_quartiles = wb.copy()
wb_quartiles['category'] = None
wb_quartiles.loc[(wb_quartiles['Gross domestic product: % growth : 2016'] < q1) | (wb_quartiles['Gross domestic product: % growth : 2016'] > q3), 'category'] = 'Outside of the middle 50%'
wb_quartiles.loc[(wb_quartiles['Gross domestic product: % growth : 2016'] > q1) & (wb_quartiles['Gross domestic product: % growth : 2016'] < q3), 'category'] = 'In the middle 50%'

sns.histplot(wb_quartiles, x="Gross domestic product: % growth : 2016", hue="category")
sns.rugplot([q1, q2, q3], c="firebrick", lw=6, height=0.1);
```

- The **Interquartile range (IQR)** is a measure of spread of a distribution

$$IQR = Q_3 - Q_1$$

<strong>Definition:</strong> **Outliers**

- 1.5 times IQR away from interquartile range (IQR)

$$Value > Q_3 + (1.5 \times IQR)$$

$$Value < Q_1 - (1.5 \times IQR)$$

## 3 Kernel Density Estimation (KDE)

- A **kernel density estimate (KDE, 核密度估计)** attempts to **approximate the underlying probability distribution from which our dataset was drawn** (To approximate a curve)

- It **estimates the probability density function of a random variable**

	- So a KDE curve will always have an **area under the curve of 1**

-  It allows us to **represent general trends** in a distribution **without focusing on the details**

<div style="background-color:#9ca3af30; padding:12px 16px; border-left:5px solid #9ca3af; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Note:</strong> **Why named Kernel?**

- 因为这里的 kernel 不是“核心/内核”那个日常意思，而是数学里一个比较专门的词：核函数

- 在 KDE 里，kernel $K(\cdot)$ 是一个“小模板函数”。你把这个模板复制到每个数据点上，再把它们叠加起来

- **kernel = 一个最基本的小组件，用它对周围的东西产生作用**

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 

warnings.filterwarnings("ignore", "use_inf_as_na") # Supresses distracting deprecation warnings

wb = pd.read_csv("world_bank.csv", index_col=0)
wb = wb.rename(columns={'Antiretroviral therapy coverage: % of people living with HIV: 2015':"HIV rate",
                       'Gross national income per capita, Atlas method: $: 2016':'gni'})
wb.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(data = wb, x = 'HIV rate', \
                       kde = True, stat = "density")

plt.title("Distribution of HIV rates");
```

- More samples, more accurate and more like a density func



### Constructing a KDE

- Place a kernel at each datapoint

- Normalize the kernels to have a total area of 1 across of kernels

- Sum the mormalized kernels

Let's see an example

- A small, artificially generated dataset of 5 datapoints: 
$[
2.2
,
2.8
,
3.7
,
5.3
,
5.7
]$

**Step 1: Place a Kernel at Each Data Point**

```python
data = [2.2, 2.8, 3.7, 5.3, 5.7]

sns.rugplot(data, height=0.3)

plt.xlabel("Data")
plt.ylabel("Density")
plt.xlim(-3, 10)
plt.ylim(0, 0.5);
```

- To begin generating a density curve, we need to choose a **kernel** and **bandwidth value (α)**

- A **kernel** is a type of density curve. It is the mathematical function that attempts to capture the randomness of each data point in our sampled data

  - Consider just one of the datapoints in our dataset: 2.2. We obtained this datapoint by randomly sampling some information out in the real world
  
  - If we were to **sample a new datapoint**, we may obtain a **slightly different** value
  
    - It could be higher than 2.2; it could also be lower than 2.2
    
	- **Assumption** that any **future sampled datapoints will be similar in value to the data we’ve already drawn**
	
  - This means that our kernel – our description of the probability of randomly sampling any new value – will be greatest at the datapoint we’ve already drawn but still have non-zero probability above and below it
    
	- The area under any kernel should integrate to 1

- A **bandwidth value**, usually denoted by α, represents the width of the kernel

  - A large value of α will result in a wide, short kernel function, while a small value with result in a narrow, tall kernel

```python
def gaussian_kernel(x, z, a):
    # We'll discuss where this mathematical formulation came from later
    return (1/np.sqrt(2*np.pi*a**2)) * np.exp((-(x - z)**2 / (2 * a**2)))

# Plot our datapoint
sns.rugplot([2.2], height=0.3)

# Plot the kernel
x = np.linspace(-3, 10, 1000)
plt.plot(x, gaussian_kernel(x, 2.2, 1))

plt.xlabel("Data")
plt.ylabel("Density")
plt.xlim(-3, 10)
plt.ylim(0, 0.5);
```

- We can create 5 kernels for our 5 datapoints

- Take their values as means for 5 distributions (Put kernels at the datapoints)

```python
def create_kde(kernel, pts, a):
    # Takes in a kernel, set of points, and alpha
    # Returns the KDE as a function
    def f(x):
        output = 0
        for pt in pts:
            output += kernel(x, pt, a)
        return output / len(pts) # Normalization factor
    return f

def plot_kde(kernel, pts, a):
    # Calls create_kde and plots the corresponding KDE
    f = create_kde(kernel, pts, a)
    x = np.linspace(min(pts) - 5, max(pts) + 5, 1000)
    y = [f(xi) for xi in x]
    plt.plot(x, y);
    
def plot_separate_kernels(kernel, pts, a, norm=False):
    # Plots individual kernels, which are then summed to create the KDE
    x = np.linspace(min(pts) - 5, max(pts) + 5, 1000)
    for pt in pts:
        y = kernel(x, pt, a)
        if norm:
            y /= len(pts)
        plt.plot(x, y)
    
    plt.show();
    
plt.xlim(-3, 10)
plt.ylim(0, 0.5)
plt.xlabel("Data")
plt.ylabel("Density")

plot_separate_kernels(gaussian_kernel, data, a = 1)
```

**Step 2: Normalize Kernels to Have a Total Area of 1**

- If we were to directly sum the kernels as they are, we would produce a KDE curve with an integrated area of $(5\ kernels) \times (area\ of\ 1\ each) = 5$

- To avoid this, we will normalize each of our kernels by multiplying each kernel by $\frac{1}{\#datapoints}$

```python
plt.xlim(-3, 10)
plt.ylim(0, 0.5)
plt.xlabel("Data")
plt.ylabel("Density")

# The `norm` argument specifies whether or not to normalize the kernels
plot_separate_kernels(gaussian_kernel, data, a = 1, norm = True)
```

**Step 3: Sum the Normalized Kernels**

```python
plt.xlim(-3, 10)
plt.ylim(0, 0.5)
plt.xlabel("Data")
plt.ylabel("Density")

plot_kde(gaussian_kernel, data, a=1)
```

- A general KDE fomula func



- $f_α(x)$ is the end result → Height of the final KDE curve at any x-value

- $K_α(x, x_i)$ is the kernel func centered on the observation i
  
  - Each kernel individually has area 1

- $n$ is the number of observed datapoints

  - We multiply by $\frac{1}{n}$ to make the total area of the KDE is 1

- Each $x_i \in \{x_1, ..., x_n\}$ represents an observed datapoint

- $\alpha$ (alpha) is the bandwidth or smoothing parameter

<div style="background-color:#9ca3af30; padding:12px 16px; border-left:5px solid #9ca3af; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Note:</strong> **Gaussian Kernel**

- There are many kernels to choose

- Generally we just use Gaussian Kernel

$$K_a(x, x_i) = \frac{1}{\sqrt{2\pi\alpha^2}} e^{-\frac{(x-x_i)^2}{2\alpha^2}}$$

</div>

- A higher $\alpha$ (alpha) entails a smoother KDE







```python
plt.xlabel("Data")
plt.xlim(-3, 10)
plt.ylim(0, 0.5)

sns.kdeplot(data, bw_method=0.65) 
sns.histplot(data, stat='density', bins=2);
```

```python
# You can also get a very similar result in a single call by requesting the KDE be added to the histogram, with kde=True and some extra keywords
plt.xlabel("Data")
plt.xlim(-3, 10)
plt.ylim(0, 0.5)

sns.histplot(data, bins=2, kde=True, stat="density", kde_kws=dict(cut=3, bw_method=0.65));
```

<div style="background-color:#9ca3af30; padding:12px 16px; border-left:5px solid #9ca3af; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Note:</strong> **Boxcar Kernel**

- Another kernel to select. It assigns a uniform density to points within a “window” of the observation, and a density of 0 elsewhere

$$K_a(x, x_i) =
\begin{cases}
\frac{1}{\alpha}, & |x-x_i| \leq \frac{\alpha}{2} \\
0, & \text{else}
\end{cases}$$

- It is **seldom used in practice** – we include it here to demonstrate that **a kernel function can take whatever form you would like**

</div>

```python
def boxcar_kernel(alpha, x, z):
    return (((x-z)>=-alpha/2)&((x-z)<=alpha/2))/alpha

xs = np.linspace(-5, 5, 200)
alpha=1
kde_curve = [boxcar_kernel(alpha, x, 0) for x in xs]
plt.plot(xs, kde_curve);
```

### Downsides of KDE

- They **make up information**: a KDE plot can make a distribution look wider, taller, or shorter than it really is

- They **hide interesting patterns**: a small peak could be an interesting pattern worth investigating, but KDE can blur it out

- They can **give a false sense of confidence**: A KDE curve based on n = 10 can appear just as smooth as one based on n = 10,000, potentially hiding the fact that we have very little data
