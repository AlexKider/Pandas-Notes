## 1 Censuses & Surveys

<div style="background-color:#3b82f630; padding:12px 16px; border-left:5px solid #3b82f6; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Definition:</strong> **Censuses 普查 & Surveys 调查**

- A **census** is a **complete count or survey of a population**, typically recording various details of individuals

- A **survey** is **a set of questions**

</div>

- Censuses are great but too expensive and impossible sometimes

  - Since it needs to survey on **every** person

- Mostly we just can get a small part of total population (lower cost)

<div style="background-color:#3b82f630; padding:12px 16px; border-left:5px solid #3b82f6; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Definition:</strong> **Sample 样本**

- A **sample** is (usually) a subset of the population that is often used to make inferences about the population

</div>

- If a sample can **properly represent the population**, we can use properties of it to infer the population's

- How the sample is frown hugely affects the inferences

**2 common sources of error**

- **Chance Error 随机误差**: random samples can vary from what is expected in any direction (即使你的抽样方法完全没问题，随机抽出来的样本也不可能每次都刚好等于总体)

- **Bias 偏差**: a systematic error in one direction (你的抽样方法本身有问题，导致结果系统性地往某个方向偏)

### Some Vocabs



- **Population / Target Population 总体**: The group that you want to learn something about which contains **individuals 个体**

- **Sampling Frame 抽样框**: The list from which the sample is drawn

  - If sampling people, then the sampling frame is the set of all people that could possibly end up in your sample
  
  - There may be individuals in your sampling frame (and hence, your sample) that are not in your population

- **Sample**: Who you actually end up sampling. The sample is therefore **a subset of your sampling frame**

**Ideally, these three should be the same**

## 2 Sampling Errors

### Selection bias

- **Systematically excludes or favors** particular groups

- e.g. The Literary Digest poll excludes people not in phone books

- How to avoid: **Randomly sample**, and **improve overlap of sampling frame and population**

### Reporting/Response/Measurement Bias

- People do not always respond truthfully, they sometimes lie

- e.g. Illegal immigrants might not answer truthfully when asked citizenship questions on the census survey

- How to avoid: Improve questions. Lots of response bias [subtypes + prevention methods](https://en.wikipedia.org/wiki/Response_bias#:~:text=%5B7%5D-,Types,-%5Bedit%5D)

### Non-response bias

- People sometimes do not complete some questions

- e.g. Only 2.4m out of 10m people responded to the Literary Digest’s poll

- How to avoid: **Increase response rate**. For example, reduce the number and length of questions, incentivize completion, and follow up

<strong>Key idea:</strong> 

- All errors & biases exists because of the **incosistency & incompletence of sampling frame & poplation & sample**

</div>

## 3 Types of Sampling

- A huge sample size does not fix a bad sampling method

- Our goal is to gather a sample that is representative of the population

### Convenience Sample

- A convenience sample is whatever you can get ahold of

- Can be non-random

- Haphazard sampling is not necessarily random sampling; there are many potential sources of bias

### Probability Sample / Random Sample

- In a probability sample, we **provide the chance that any specified set of individuals will be in the sample** 

- Individuals in the population can have different chances of being selected; they don’t all have to be uniform and we sample at random based off this known chance

<div style="background-color:#8b5cf630; padding:12px 16px; border-left:5px solid #8b5cf6; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Question:</strong> **Why Randomness?**

- Since we know the source probabilities (distributions), we can measure the errors, estimate the bias (quantify the uncertainty)

- Sampling randomly can reduce bias and entail a more representative sample

	- This is true **only if the distribution we choose is accurate**

</div>

### Common Random Sampling Schemes

- **Uniform random sample with replacement**

- **Simple random sample (SRS)**

  - Drown uniformly
  
  - Without replacement

- **Stratified random sample**

  - The sample frame is devided into **non-overlapping strata**
  
  - Then conduct **SRS on each stratum**
  
  - **Sample size is proportional to stratum size**

## 4 Non-representative Sampling & Post-Stratification

### Non-representative Sampling

- Sometimes we just cannot obtain a truly reprsentative sample

	**e.g.**

- When conducting customer surveys, not everyone responds or provides contact info


### Post-Stratification

- Divide **both the sample & population** into distinct cells according to chosen characteristics

- Compute the overall response in each sample cell

- Aggregate over the sample cells, weighing each cell proportionally by the size of **corresponding population cell**

<div style="background-color:#eab30826; padding:12px 16px; border-left:5px solid #eab308; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Key idea:</strong> **Assumptions**

- In order to do this, we must have:

  - The population cell sizes are accurate
  
  - Each sample cell is representative of the corresponding population cell

</div>

**e.g.**

**Actual population**

- 8000 undergrads (80%), 2000 masters (20%)

**The survey you get**

- 50 undergrads, 50 masters (50% for both)
	
- 70% of undergrad surveys like studying, 30% of masters like studying
  
Actual prob. of like studying is $80\% \times 70\% + 20\% \times 30\% = 62\%$

## 5 *Demo: The 1936 Presidential Election Results (exer.)

- Each row of votes represents a single voter

- voted_dem is 0/1 variable:

  - **1**: Franklin D. Roosevelt, Democratic candidate
  
  - **0**: Alf Landon, Republican candidate

```python
import zipfile
import pandas as pd

with zipfile.ZipFile("1936_votes.zip", 'r') as z:
    with z.open("1936_votes.csv") as csv_file:
        votes = pd.read_csv(csv_file)

votes.head()
```

- How many votes voted?

```python
len(votes)
```

- What fraction of voters voted for Roosevelt, the Democratic candidate?

```python
votes['voted_dem'].sum() / len(votes)
```

- Don’t the slides say that Roosevelt won 61% of the vote?

- Yes! But, he won 61% of all votes. If we filter to just the voters who voted for Roosevelt or Landon, Roosevelt won 62.5% of votes

```python
votes['voted_dem'].mean()
```

- Of the 44,430,549 voters who voted for either Roosevelt or Landon, 62.5% voted for Roosevelt

### SRS

<div style="background-color:#9ca3af30; padding:12px 16px; border-left:5px solid #9ca3af; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Note:</strong> **Name in a hat**

- An SRS is sometimes called a “names in a hat” sample, since it’s a lot like putting each observation on a slip of paper, putting all the slips in a big hat, and then randomly drawing slips one at a time

</div>

- We take a simple random sample of just 1,000 voters and calculate the proportion who planned to vote for Roosevelt, how close would we be to 62.5%?

```python
votes['voted_dem'].sample(1000).mean()

# It changes every time you run it because it is random sampling
# You can set a random seed to make it reproducible
```

- Note that the cell above is a little slow, since we’re sampling from a DataFrame with almost 45 million rows

- We can speed up the sampling using NumPy

```python
import numpy as np

# Construct a random number generator object.
# No need to be familiar with using NumPy this way in Data 100!
rng = np.random.default_rng()

n_votes = len(votes)

# Generate 1000 random integers from 0 to (number of votes - 1)
idx = rng.integers(low=0, high=n_votes-1, size=1000)

votes['voted_dem'].iloc[idx].mean()
```

- They are all close to 62.5%

```python
for _ in range(10):
  idx = rng.integers(low=0, high=n_votes-1, size=1000)
  print(votes['voted_dem'].iloc[idx].mean())
```

- Let’s randomly generate 10,000 estimates

```python
nrep = 10000   # number of simulations
n = 1000       # size of our sample
results = []   # list to store the sampling results

for i in range(0, nrep):
    idx = rng.integers(low=0, high=n_votes, size=1000)
    results.append(votes['voted_dem'].iloc[idx].mean())

# First 10 simulated sample proportions
results[:10]
```

- Plotting our estimates with KDE

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 3))
p = sns.histplot(x=results, kde=True, bins=10)

# Make x-axis centered at 0.625 with 0.01 intervals
p.set_xticks(np.arange(0.625 - 5 * 0.01, 0.625 + 5 * 0.01, 0.01));
```
