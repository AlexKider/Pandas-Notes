## 1. Tabular Data and pandas

<strong>Definition:</strong> **Tables**



- The blue box is one observation
- The yellow box is one feature

**Tables** allow us to represent each **observation**, or instance of collecting data from an individual, as its own row

We can record each observation’s distinct characteristics, or **features**, in separate columns

</div>

```python
# `pd` is the conventional alias for Pandas, as `np` is for NumPy
import pandas as pd
```

### Three Fundamental Data Structures



**Series**: 1D labeled array data; best thought of as columnar data

**DataFrame**: 2D tabular data with rows and columns

**Index**: A sequence of row/column labels

- Both contain an Index, or a shared list of row labels (the integers from 0 to 4, inclusive)

### Series

#### Basics

```python
s = pd.Series(["welcome", "to", "data 100"])
s
```

```python
# Accessing data values within the Series
s.values
```

```python
# Accessing the Index of the Series
s.index
```

By default, the index of a Series is a sequential list of integers beginning from 0

Optionally, a manually specified list of desired indices can be passed to the index argument

```python
s = pd.Series([-1, 10, 2], index = ["a", "b", "c"]) # set your own index
s
```

Indices can also be changed

```python
s.index = ["first", "second", "third"]
s
```

#### Selection in Series

**There are three primary methods**

- A single label
- A list of labels
- A filtering condition

```python
s = pd.Series([4, -2, 0, 6], index = ["a", "b", "c", "d"])
s
```

```python
# Single label
# We return the value stored at the index label "a"
s["a"] # Just one data point as index!
```

```python
# List of labels
# We return a Series of the values stored at the index labels "a" and "c"
s[["a", "c"]] # Since we have multiple indices, now it is not a data point, it is a iterables
```

<div style="background-color:#eab30826; padding:12px 16px; border-left:5px solid #eab308; width:100%; max-width:100%; box-sizing:border-box; overflow-x:auto;">

<strong>Key idea:</strong>

In order to use filters, we need to create a new Series of booleans

</div>

```python
# Filter condition: select all elements greater than 0
s > 0 
```

We then use this boolean condition to index into our original Series. pandas will select only the entries in the original Series that satisfy the condition

```python
s[s > 0] 
```
